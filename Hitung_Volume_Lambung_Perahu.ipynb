```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
```

Panjang Kapal
$$
L = 4\ m
$$

Pada sketsa perahu, satu interval atau satu kotak grid terdiri dari dua pola dash-kosong, maka $$
1\text{ interval}=2(0,1)=0,2\text{ m}
$$

Jumlah interval sepanjang kapal pada gambar sebesar 20 interval, sehingga
$$
L=20(0,2)=4\text{ m}
$$

Diperoleh, panjang kapal yang digunakan adalah 4 m

**Tampak Atas**

```python
ukuran_dash_cm = 5
ukuran_celah_cm = 5

ukuran_dash_kosong_cm = ukuran_dash_cm + ukuran_celah_cm
ukuran_dash_kosong_m = ukuran_dash_kosong_cm / 100

# Satu kotak grid
dash_kosong_per_interval = 2

interval_m = (
    dash_kosong_per_interval *
    ukuran_dash_kosong_m
)

# Jumlah interval pada perahu
jumlah_interval = 20

# Panjang kapal pada perahu
panjang_kapal = jumlah_interval * interval_m

print("Ukuran 1 dash-kosong =", ukuran_dash_kosong_m, "m")
print("Ukuran 1 interval grid =", interval_m, "m")
print("Jumlah interval =", jumlah_interval)
print("Panjang kapal =", panjang_kapal, "m")
```

**Hasil Tampak Atas dan Tampak Samping**

```python
# Jumlah interval tampak atas
jumlah_interval_atas = jumlah_interval

titik_atas = np.arange(jumlah_interval_atas + 1)

# Jarak antar titik
interval_m_atas = interval_m

# Posisi setiap titik
jarak_atas = titik_atas * interval_m_atas

# Lebar lambung pada tampak atas
lebar_atas = np.array([
    0.45, 0.60, 0.70, 0.75, 0.80,
    0.80, 0.80, 0.80, 0.80, 0.80,
    0.80, 0.80, 0.80, 0.80, 0.80,
    0.75, 0.70, 0.60, 0.50, 0.35,
    0.15
])

# Lebar satu lambung
lebar_satu_lambung = lebar_atas / 2

# Tinggi dari tampak samping
tinggi_atas = tinggi

hasil_tinggi_atas = pd.DataFrame({
    "Titik": titik_atas,
    "Jarak (m)": jarak_atas,
    "Lebar satu lambung (m)": lebar_satu_lambung,
    "Tinggi (m)": tinggi_atas
})

hasil_tinggi_atas["Jarak (m)"] = hasil_tinggi_atas["Jarak (m)"].round(2)
hasil_tinggi_atas["Lebar satu lambung (m)"] = hasil_tinggi_atas["Lebar satu lambung (m)"].round(2)
hasil_tinggi_atas["Tinggi (m)"] = hasil_tinggi_atas["Tinggi (m)"].round(3)

hasil_tinggi_atas
```

Rumus Luas Penampang:
$$
A_i = b_i \times h_i
$$

Keterangan:
- $A_i$ = luas penampang pada titik ke-$i$ ($m^2$)
- $b_i$ = lebar lambung pada titik ke-$i$ (m)
- $h_i$ = tinggi lambung pada titik ke-$i$ (m)

```python
# Hitung luas penampang
hasil_tinggi_atas["Luas Penampang (m2)"] = (
    hasil_tinggi_atas["Lebar satu lambung (m)"] *
    hasil_tinggi_atas["Tinggi (m)"]
)

hasil_tinggi_atas["Luas Penampang (m2)"] = (
    hasil_tinggi_atas["Luas Penampang (m2)"].round(4)
)

hasil_tinggi_atas
```

**Hitung Volume Lambung Kapal**

**Aturan Simpson 1/3**

Rumus: $$
V = \frac{\Delta x}{3}
\left[
A_0 + A_n
+ 4(A_1 + A_3 + A_5 + \cdots + A_{n-1})
+ 2(A_2 + A_4 + A_6 + \cdots + A_{n-2})
\right]
$$

Keterangan:
- $V$ = volume satu lambung ($m^3$)
- $\Delta x$ = jarak antar titik (m)
- $A_0$ = luas penampang awal
- $A_n$ = luas penampang akhir
- Luas penampang dengan indeks ganjil dikalikan 4
- Luas penampang dengan indeks genap dikalikan 2

```python
# data luas penampang
luas = hasil_tinggi_atas["Luas Penampang (m2)"].values

# aturan Simpson 1/3
volume_satu_lambung = (
    interval_m_atas / 3 *
    (
        luas[0]
        + luas[-1]
        + 4 * luas[1:-1:2].sum()
        + 2 * luas[2:-1:2].sum()
    )
)

volume_satu_lambung = round(volume_satu_lambung, 4)
print("Volume satu lambung =", volume_satu_lambung, "m3")
```

```python
# Volume kedua lambung
volume_kedua_lambung = 2 * volume_satu_lambung

print(
    "Volume kedua lambung =",
    volume_kedua_lambung,
    "m3"
)
```

**Aturan Trapesium**

Rumus: $$
V = \Delta x
\left[
\frac{A_0 + A_n}{2}
+ \sum_{i=1}^{n-1} A_i
\right]
$$

Keterangan:
- $V$ = volume satu lambung ($m^3$)
- $\Delta x$ = jarak antar titik atau interval (m)
- $A_0$ = luas penampang pada titik awal ($m^2$)
- $A_n$ = luas penampang pada titik akhir ($m^2$)
- $A_i$ = luas penampang pada titik ke-$i$ ($m^2$)
- $\sum_{i=1}^{n-1} A_i$ = jumlah seluruh luas penampang di antara titik awal dan titik akhir


```python
# aturan trapesium
volume_trapesium = (
    interval_m_atas *
    (
        (luas[0] + luas[-1]) / 2
        + luas[1:-1].sum()
    )
)

print("Volume satu lambung - Trapesium =", volume_trapesium, "m3")
```

```python
# volume kedua lambung
volume_kedua_trapesium = 2 * volume_trapesium

print("Volume kedua lambung - Trapesium =",volume_kedua_trapesium,"m3")
```

**Hasil Perbandingan Antara kedua Aturan**

```python
hasil_akhir = pd.DataFrame({
    "Aturan": [
        "Simpson 1/3",
        "Trapesium"
    ],
    "Volume 1 Lambung (m3)": [
        volume_satu_lambung,
        volume_trapesium
    ],
    "Volume 2 Lambung (m3)": [
        volume_kedua_lambung,
        volume_kedua_trapesium
    ]
})

hasil_akhir["Volume 1 Lambung (m3)"] = hasil_akhir["Volume 1 Lambung (m3)"].round(5)
hasil_akhir["Volume 2 Lambung (m3)"] = hasil_akhir["Volume 2 Lambung (m3)"].round(5)

# konveri ke liter
hasil_akhir["Volume 1 Lambung (liter)"] = (hasil_akhir["Volume 1 Lambung (m3)"] * 1000).round(2)
hasil_akhir["Volume 2 Lambung (liter)"] = (hasil_akhir["Volume 2 Lambung (m3)"] * 1000).round(2)

hasil_akhir
```
