**Gabungan Tampak Atas dan Tampak Samping**

```python
# Jumlah interval tampak atas
jumlah_interval_atas = 20

titik_atas = np.arange(jumlah_interval_atas + 1)

# Jarak antar titik = 4 / 20 = 0.2 m
interval_m_atas = 4.0 / jumlah_interval_atas

# Posisi setiap titik
jarak_atas = titik_atas * interval_m_atas

# Lebar lambung tampak atas
lebar_atas = np.array([
    0.45, 0.60, 0.70, 0.75, 0.80,
    0.80, 0.80, 0.80, 0.80, 0.80,
    0.80, 0.80, 0.80, 0.80, 0.80,
    0.75, 0.70, 0.60, 0.50, 0.35,
    0.15
])

# Posisi tampak samping
x_samping = jarak_samping

# ke posisi titik tampak atas
tinggi_atas = np.interp(
    jarak_atas,
    x_samping,
    tinggi_samping
)

hasil_tinggi_atas = pd.DataFrame({
    "Titik": titik_atas,
    "Jarak (m)": jarak_atas,
    "Lebar (m)": lebar_atas,
    "Tinggi (m)": tinggi_atas
})

# Membulatkan untuk tampilan
hasil_tinggi_atas["Jarak (m)"] = hasil_tinggi_atas["Jarak (m)"].round(2)
hasil_tinggi_atas["Lebar (m)"] = hasil_tinggi_atas["Lebar (m)"].round(2)
hasil_tinggi_atas["Tinggi (m)"] = hasil_tinggi_atas["Tinggi (m)"].round(3)

hasil_tinggi_atas
```

```python
# hitung luas penampang
hasil_tinggi_atas["Luas Penampang (m2)"] = (
    hasil_tinggi_atas["Lebar (m)"] *
    hasil_tinggi_atas["Tinggi (m)"]
)

hasil_tinggi_atas["Luas Penampang (m2)"] = (
    hasil_tinggi_atas["Luas Penampang (m2)"].round(4)
)

hasil_tinggi_atas
```

**Hitung Volume Lambung Kapal**

**Aturan Simpson 1/3**

Rumus:

```python
# data luas penampang
luas = hasil_tinggi_atas["Luas Penampang (m2)"].values

# aturan Simpson 1/3
volume_satu_lambung = (
    interval_m_atas / 3 *
    (
        luas[0]
        + luas[-1]
        + 4 * luas[1:-1:2].sum()
        + 2 * luas[2:-1:2].sum()
    )
)

volume_satu_lambung = round(volume_satu_lambung, 4)
print("Volume satu lambung =", volume_satu_lambung, "m3")
```

```python
# Volume kedua lambung
volume_kedua_lambung = 2 * volume_satu_lambung

print(
    "Volume kedua lambung =",
    volume_kedua_lambung,
    "m3"
)
```

**Aturan Trapesium**

```python
# aturan trapesium
volume_trapesium = (
    interval_m_atas *
    (
        (luas[0] + luas[-1]) / 2
        + luas[1:-1].sum()
    )
)

print("Volume satu lambung - Trapesium =",
      volume_trapesium, "m3")
```

```python
# volume kedua lambung
volume_kedua_trapesium = 2 * volume_trapesium

print(
    "Volume kedua lambung - Trapesium =",
    volume_kedua_trapesium,
    "m3"
)
```

**Hasil Perbandingan Antara kedua Aturan**

```python
hasil_akhir = pd.DataFrame({
    "Aturan": [
        "Simpson 1/3",
        "Trapesium"
    ],
    "Volume 1 Lambung (m3)": [
        volume_satu_lambung,
        volume_trapesium
    ],
    "Volume 2 Lambung (m3)": [
        volume_kedua_lambung,
        volume_kedua_trapesium
    ]
})

hasil_akhir["Volume 1 Lambung (m3)"] = hasil_akhir["Volume 1 Lambung (m3)"].round(5)
hasil_akhir["Volume 2 Lambung (m3)"] = hasil_akhir["Volume 2 Lambung (m3)"].round(5)

hasil_akhir
```
